```python
import pandas as pd
import numpy as np

# Combine all symptom columns into a single Series of lists
all_symptoms = df[['Symptom_1', 'Symptom_2', 'Symptom_3']].values.tolist()

# Flatten the list of lists and get unique symptoms
flat_list = [item for sublist in all_symptoms for item in sublist if pd.notna(item)]
unique_symptoms = list(set(flat_list))

# Create new columns for one-hot encoding
for symptom in unique_symptoms:
    df[symptom] = 0

# Populate the one-hot encoded columns
for index, row in df.iterrows():
    symptoms_in_row = [row['Symptom_1'], row['Symptom_2'], row['Symptom_3']]
    for symptom in symptoms_in_row:
        if pd.notna(symptom) and symptom in unique_symptoms:
            df.at[index, symptom] = 1

# Drop the original symptom columns if desired
# df = df.drop(['Symptom_1', 'Symptom_2', 'Symptom_3'], axis=1)

# Display the updated DataFrame with one-hot encoding
print(df.head())
```

```python
import pandas as pd
# Perform one-hot encoding on the 'Diagnosis' column
diagnosis_one_hot = pd.get_dummies(df['Diagnosis'], prefix='Diagnosis')

# Convert the boolean values (True/False) to integers (1/0)
diagnosis_one_hot = diagnosis_one_hot.astype(int)

# Join the new one-hot encoded columns back to the original DataFrame
df = df.join(diagnosis_one_hot)

# Display the updated DataFrame
df.columns
```

```python
import pandas as pd
# Perform one-hot encoding on the 'Treatment_Plan' column
treatment_plan_one_hot = pd.get_dummies(df['Treatment_Plan'], prefix='Treatment_Plan')

# Convert the boolean values (True/False) to integers (1/0)
treatment_plan_one_hot = treatment_plan_one_hot.astype(int)

# Join the new one-hot encoded columns back to the original DataFrame
df = df.join(treatment_plan_one_hot)

# Display the updated DataFrame
print(df.head())
df.columns
```

```python
# prompt: want to do corr() so i need to pre process Blood_Pressure_mmHg column that has values such as '132/91' and '109/106' to columns systolic and average

import pandas as pd
# Function to extract systolic and calculate average blood pressure
def process_blood_pressure(bp_str):
    if isinstance(bp_str, str) and '/' in bp_str:
        try:
            systolic, diastolic = map(int, bp_str.split('/'))
            average = (systolic + diastolic) / 2
            return systolic, average
        except ValueError:
            return None, None  # Handle cases where conversion to int fails
    return None, None # Handle cases where format is not as expected

# Apply the function to create new columns
df[['Systolic_BP', 'Average_BP']] = df['Blood_Pressure_mmHg'].apply(lambda x: pd.Series(process_blood_pressure(x)))

# Display the updated DataFrame with new columns
print(df[['Blood_Pressure_mmHg', 'Systolic_BP', 'Average_BP']].head())
```

```python
# prompt: compare the values of new_case and values of entries for case_base to find the case in case_base most similar to new_case. calculate with neighbor similarity

import pandas as pd
def weighted_neighbor_similarity(case1, case2, weights):
    """
    Calculates the weighted neighbor similarity between two cases.

    Args:
        case1 (pd.Series): The first case (as a pandas Series).
        case2 (pd.Series): The second case (as a pandas Series).
        weights (dict): A dictionary where keys are feature names and values
                        are their weights.

    Returns:
        float: The weighted neighbor similarity score.
    """
    intersection = 0
    union = 0
    for feature in weights:
        weight = weights[feature]
        if case1[feature] == 1 and case2[feature] == 1:
            intersection += weight
        if case1[feature] == 1 or case2[feature] == 1:
            union += weight

    # Avoid division by zero if union is 0
    if union == 0:
        return 0
    return intersection / union

# Calculate weights based on rarity (or any other desired method)
# For simplicity, let's use a basic weight based on the feature's presence
# in the training data. You can replace this with your rarity calculation.
feature_weights = {}
for col in features:
  # Simple weight: 1 if present in at least one training case, 0 otherwise
  feature_weights[col] = 1 if case_base[col].sum() > 0 else 0

# Ensure weights are calculated for all relevant features
# You might want to use your rarity_df for a more sophisticated weighting
# For demonstration, let's use a simple equal weight for all features in `features`
feature_weights = {feature: 1 for feature in features}


# Initialize variables to store the best similarity and the most similar case
best_similarity = -1
most_similar_case_index = -1

# Iterate through each case in the case_base
for index, case_entry in case_base[features].iterrows():
    # Calculate the weighted neighbor similarity between the new case and the current case entry
    similarity = weighted_neighbor_similarity(new_case.iloc[0][features], case_entry, feature_weights)

    # Update the best similarity and the index of the most similar case if the current similarity is better
    if similarity > best_similarity:
        best_similarity = similarity
        most_similar_case_index = index

# Get the most similar case entry from the original DataFrame (or case_base)
most_similar_case = case_base.iloc[most_similar_case_index]

print(f"\nMost similar case found at index: {most_similar_case_index}")
print(f"Weighted Neighbor Similarity: {best_similarity:.4f}")
print("\nDetails of the most similar case:")
print(most_similar_case)

# You can also retrieve the original row from the 'df' DataFrame
# using the index if you need more columns than what's in case_base
original_most_similar_row = df.iloc[most_similar_case_index]
print("\nOriginal details from the full dataset:")
original_most_similar_row
```

```python
from sklearn.naive_bayes import GaussianNB

X_nb = case_base[symptoms]

treatment_plan_models = {} #menampung model untuk setiap jenis perawatan

for treatment_col in treatment_plan_cols:
  y_nb = case_base[treatment_col]

  nb_model = GaussianNB()
  nb_model.fit(X_nb, y_nb)
  treatment_plan_models[treatment_col] = nb_model

new_case_symptoms = new_case[symptoms]

treatment_plan_probabilities = {}#menampung nilai kepercayaan prediksi model setiap jenis perawatan
for treatment_col, model in treatment_plan_models.items():
  treatment_plan_probabilities[treatment_col] = model.predict_proba(new_case_symptoms)[:, 1][0]

print("\nNilai kepercayaan Naive Bayes kasus baru untuk setiap jenis perawatan :")
print(treatment_plan_probabilities)

best_treatment_plan_col = max(treatment_plan_probabilities, key=treatment_plan_probabilities.get)
highest_probability = treatment_plan_probabilities[best_treatment_plan_col]

print(f"\nBest Treatment Plan based on Naive Bayes probability: {best_treatment_plan_col}")
print(f"Highest Probability: {highest_probability:.4f}")

if best_treatment_plan_col in new_case.columns:
    new_case[best_treatment_plan_col] = 1 # Set the recommended treatment plan to 1
else:
    print(f"Warning: Recommended treatment plan '{best_treatment_plan_col}' not found in new_case columns.")


print("\nNew case DataFrame after Naive Bayes recommendation:")
new_case[treatment_plan_cols]
```
